```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tkinter as tk
```

```python
df = pd.read_csv('../dataset/Salary_dataset.csv')
```

```python
print('Kolom pada dataset :', df.shape[1])
print('Baris pada dataset :', df.shape[0])
df.head()
```

```python
df.info()
df.describe()
```

```python
nilai_kosong = df.isnull().sum()

pd.DataFrame({
    'Nilai Kosong': nilai_kosong,
    'persentase nilai kosong': (nilai_kosong / len (df))*100
    
})
```

```python
nilai_duplikat = df.duplicated().sum()

pd.DataFrame({
    'Nilai Duplikat': [nilai_duplikat],
    'persentase nilai duplikat': [(nilai_duplikat / len (df))*100]
    
})
```

```python
cols = df.select_dtypes(include=['number'])

Q1 = cols.quantile(0.25)
Q3 = cols.quantile(0.75)

IQR = Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier = ((cols < batas_bawah) | (cols > batas_atas)).sum()
outlier
```

```python
cols = df.drop(columns=['id'])

plt.figure(figsize=(15, 12))

for c, col in enumerate[any](cols, start=1):
    plt.subplot(3, 3, c)
    sns.boxplot(data=(df[col]))
    plt.title('')

plt.show()
```

```python
plt.figure(figsize=(15, 12))

for c, col in enumerate[any](cols, start=1):
    plt.subplot(3, 3, c)
    sns.histplot(data=(df[col]), kde=True)
    plt.title(col)

plt.show()
```
