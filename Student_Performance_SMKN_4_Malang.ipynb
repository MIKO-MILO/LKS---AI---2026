```python
# Mengimportkan Pandas, numpy, matplotlib dan seaborn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data set

df = pd.read_csv('dataset/student_performance.csv')
```

```python
df['class_number'] = df['class_number'].fillna(df['class_number'].mean())
```

```python
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())
```

```python
df['last_exam_score'] = df['last_exam_score'].fillna(df['last_exam_score'].mean())
```

```python
df['course_name'] = df['course_name'].fillna(df['course_name'].mode()[0])
```

```python
df.duplicated().sum()
```

```python
plt.hist(df['exam_score'], bins=10)
plt.xlabel('Nilai Ulangan')
plt.ylabel('Jumlah Data')
plt.title('View Histogram Exam Score')
plt.show()
```

```python
sns.histplot(df['exam_score'])
plt.title('Hisplot untuk Exam Score')
plt.show()
```

```python
sns.scatterplot(data=df, x='exam_score', y='study_hours')
plt.title('Scatter Plot')
plt.show()
```
