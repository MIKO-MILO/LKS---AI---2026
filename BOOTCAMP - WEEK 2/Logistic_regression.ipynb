```python
# kolom = ['age', 'cigsPerDay', 'totChol', 'sysBP', 'BMI', 'heartRate', 'glucose']

kolom = df.select_dtypes(include='float64').columns

Q1 = df[kolom].quantile(0.25)
Q3 = df[kolom].quantile(0.75)

IQR =  Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outlier = ((df[kolom] < batas_bawah) | (df[kolom] > batas_atas)).sum()

hasil = pd.DataFrame({'Jumlah Outlier': outlier, 'Persentase Outlier': outlier / len(df) * 100})
hasil = hasil.sort_values(by='Jumlah Outlier', ascending=False)
hasil
```

```python
for col in kolom:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    bb = q1 - 1.5 * iqr
    ba = q3 + 1.5 * iqr
    df[col] = df[col].clip(lower=bb, upper=ba)

# verifikasi pakai daftar yang SAMA (kolom), bukan cols
Q1 = df[kolom].quantile(0.25)
Q3 = df[kolom].quantile(0.75)
IQR = Q3 - Q1
outlier = ((df[kolom] < (Q1 - 1.5 * IQR)) | (df[kolom] > (Q3 + 1.5 * IQR))).sum()

hasil = pd.DataFrame({'Jumlah Outlier': outlier}).sort_values(by='Jumlah Outlier', ascending=False)
hasil
```

## **5. Modelling** 

```python
X = df.drop(columns=['TenYearCHD','diaBP','currentSmoker'])
y = df[['TenYearCHD']]
```

```python
X_cek = X.loc[:, X.std() > 0]
invers_cek = np.linalg.inv(X_cek.corr().values)

cek_vif = pd.DataFrame ({
    'Fitur' : X_cek.columns,
    'vif' : np.diag(invers_cek)
}).sort_values(by='vif',ascending=False).reset_index(drop=True)

cek_vif
```

```python
def split_data(X, y, train_size=0.8, random_state=42):
    np.random.seed(random_state)
    idx = np.random.permutation(len(X))
    split_ratio = int(train_size * len(X))
    X_train, X_test = X.iloc[idx[:split_ratio]], X.iloc[idx[split_ratio:]]
    y_train, y_test = y.iloc[idx[:split_ratio]], y.iloc[idx[split_ratio:]]
    return X_train, X_test, y_train, y_test

X_train, X_test, y_train, y_test = split_data(X, y)
```

```python
mean = np.mean(X_train, axis=0)
std = np.std(X_train, axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

X_train = np.array(X_train)
X_test = np.array(X_test)
y_train = np.array(y_train).reshape(-1)
y_test = np.array(y_test).reshape(-1)

m,n = X_train.shape
```

```python
def sigmoid(z):
    return 1 /(1+np.exp (-z))
```

```python
def cost_function(X,y,w,b):
    m, n = X.shape
    cost_sum = 0
    for i in range (m):
        z =np.dot(w,X[i]) + b
        g = sigmoid(z)

        cost_sum += -y[i] * np.log(g) - (1 - y[i]) * np.log(1-g)
    return (1/m) * cost_sum
```

```python
def gradient_function(X,y,w,b):
    m, n = X.shape
    grad_w = np.zeros(n)
    grad_b = 0

    for i in range (m):
        z =np.dot(w,X[i]) + b
        g = sigmoid(z)

        grad_b += (g-y[i])
        for j in range(n):
            grad_w[j] += (g - y[i]) * X[i, j]

    grad_b = (1/m) * grad_b
    grad_w = (1/m) * grad_w

    return grad_b, grad_w
```

```python
def gradient_desc(X,y,alpha,iterasi):
    m, n = X.shape
    w = np.zeros(n)
    b = 0

    for i in range (iterasi):
        grad_b,grad_w = gradient_function(X,y,w,b)

        w = w - alpha * grad_w
        b = b - alpha * grad_b

        if i % 1000 == 0:
            print(f"iterasi{i}: cost{cost_function(X,y,w,b)}")

    return w,b
```

```python
def predict(X,w,b):
    m, n = X.shape
    preds = np.zeros(m)

    for i in range(m):
        z = np.dot(w,X[i]) + b
        g = sigmoid(z)

        preds[i] = 1 if g >= 0.5 else 0

    return preds
```

## **6.0 Evaluasi Model**

```python
learning_rate = 0.001
iterasi = 1000

final_w, final_b = gradient_desc(X_train,y_train,learning_rate,iterasi)

prediksi = predict(X_train, final_w,final_b)
akurasi =  np.mean(prediksi == y_train) * 100
print(akurasi)
```

```python
y_pred = predict(X_test, final_w, final_b)
y_true = y_test

TP = np.sum((y_true == 1) & (y_pred == 1))
TN = np.sum((y_true == 0) & (y_pred == 0))
FP = np.sum((y_true == 0) & (y_pred == 1))
FN = np.sum((y_true == 1) & (y_pred == 0))

presisi = (TP / (TP + FP)) * 100 if (TP + FP) > 0 else 0
recall = (TP / (TP + FN)) * 100 if (TP + FN) > 0 else 0
f1 = (2 * presisi * recall) / (presisi + recall) if (presisi + recall) > 0 else 0


print(f"Precision : {presisi:}")
print(f"Recall    : {recall:}")
print(f"F1-Score  : {f1:}")
```

```python
# Simpan model untuk dipakai oleh GUI (gui_logistic.py)
np.savez(
    'model_logistic.npz',
    mean=np.asarray(mean, dtype=float),
    std=np.asarray(std, dtype=float),
    w=final_w,
    b=final_b,
    features=np.array(X.columns)
)
print('Model tersimpan: model_logistic.npz')
```
