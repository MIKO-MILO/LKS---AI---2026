### **Type Data**

### **1. INTEGER**

Type data integer adalah **Bilangan Bulat**

Sepeerti 1, 2, 3, 0, -1, -2, -3

```python
data = 8
print(type(data))
```

### **2. FLOAT**

Type data Float adalah **Bilangan Desimal**

Sepeerti 1.0, 2.2, 3.3, 0.5

```python
data = 8.0 
print(type(data))

data= 8.3
print(type(data))
```

### **3. BOOLEANj**

Type data Boolean adalah **data biner** / hanya memiliki 2 nilai

Sepeerti 1.0, 2.2, 3.3, 0.5

```python
data =  False #Penulisan Wajib "False" Huruf F Kapital tidak boleh "false" atau "FALSE"
data2 = True #Penulisan Wajib "True" Huruf T Kapital tidak boleh "true" atau "TRUE"

print(type(data))
print(type(data2))
```

### **4. STRING**

Type data String adalah **kumpulan karakter**

"" --> Kutip

```python
data1 = "spiderman"
data2 = "8"
data3 = "8.0"
data4 = "False"

print(type(data1))
print(type(data2))
print(type(data3))
print(type(data4))
```

### **5. LIST**

Type data List adalah **kumpulan Element** yang dapat di **MODIFIKASI/UBAH**

> Format : [10, 8.0, "element2", False]

```python
data = ["spiderman", "iroman", "superman"]
data2 = [10, 8.0, "element2", False]

print(type(data))
print(type(data2))
```

### **6. TUPLE**

Type data Tuple adalah **kumpulan Element** yang ***TIDAK*** dapat di **MODIFIKASI/UBAH**

> Format : (10, 8.0, "element2", False)

```python
data = (10, 8.0, "element2", False)

print(type(data))
```

### **7. SET**

Type data SET adalah **Kumpulan Element** yang pasti **Unik dan Tidak Berurutan**

> Format : {element1, element2, element3}

```python
data = {1, 4, 4, 4, 2, 3}

print(data)
print(type(data))
```

### **8. DICTIONARY**

Type data Dictionary set adalah Kumpulan **Pasangan Key-Value/Kunci-Nilai**

> Format : {key1:value1, key2:value2, dst}


```python
data_dic = {"eat": "makan", "run": "berlari", "sad": "sedih"}

print(data_dic)
print(type(data_dic))
```
