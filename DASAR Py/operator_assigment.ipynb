OPERATOR ASSIGMENT

Operator assigment dalam python digunakan untuk menginisialisasi atau mengubah nilai variabel

Assigment =  Penugasan

x = 5 Variabel x diinisialisasi dengan nilai 5

x += 3 Ini akan menambahkan 3 ke nilai 'x'

```python
x = 5
x = x + 2
print(x)
```

```python
x = 5
x += 2
print(x)
```

```python
x = 5
print(x)
```
