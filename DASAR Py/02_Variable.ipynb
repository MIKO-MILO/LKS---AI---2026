VARIABEL

Variable dalam python adalah tempat menyimpan nilai yang dapat diakses dan dimanipulasi pada program

```python
nama = 'piter'
print(nama)
```

```python
spiderman = 'piterparker'
print(spiderman)
```

```python
nama, alamat, jeniskelamin = 'piter parker', 'Indonesia', 'Laki-laki'

print(nama)
print(alamat)
print(jeniskelamin)
```
