```python
# type integer atau bilangan bulat

data = 8 
print(data)
print(type(data))
```

```python
# type float atau bilangan desimal

data = 8.0
print(data)
print(type(data))
```

```python
# type boolean atau biner

data1 = True
data2 = False

print(data1)
print(data2)
print(type(data1))
print(type(data2))
```

```python
# type string atau kumpulan karakter

data1 = "spiderman"
data2 = "8.0"

print (data1)
print(data2)

print(type(data1))
print(type(data2))
```

```python
# type data list

data = ["spiderman", "ironman", "superman"]

print(data)
print(type(data))
```

```python
# tuple

data = ("spiderman", "ironman", "superman")

print(data)
print(type(data))
```

```python
#type data Set

data = {"spiderman", "ironman", "superman"}

print(data)
print(type(data))
```

```python
#type data dictionary

data = {"spiderman" : "marvel", "batman": "dc"}

print(data)
print(type(data))
```
