### **COMMENT PYTHON**

Comment adalah cara untuk menambahkan catatan atau keterangan pada program yang tidak akan dieksekusi oleh Python. Biasanya comment digunakan untuk 

- Bagaimana program berkerja
- Tujuan dari program
- hal- hal penting lainnya

Comment sendiri dibagi menjadi dua yaitu :

> Single line Comment

    # ini adalah comment sigle line

> Multi-line comment

    ` """

    Program Ini adalah contoh multiline comment
    Program Ini adalah contoh multiline comment

    `"""

```python
# Ini adalah single line comment dalam bahasa Python.


"""
Dan ini adalah mutlti line comment dalam bahasa Python.
Dan ini adalah mutlti line comment dalam bahasa Python.
Dan ini adalah mutlti line comment dalam bahasa Python.
Dan ini adalah mutlti line comment dalam bahasa Python.
Dan ini adalah mutlti line comment dalam bahasa Python.
"""
```

### **Tujuan Utama Comment**

Tujuan utama dari penulisan comment dalam kode program adalah untuk memberikan penjelasan atau dokumentasi tentang kode program yang ditulis, agar orang lain yang membaca kode program tersebut dapat memahaminya dengan mudah

```python
"""
Ini merupakan contoh program pemebelajatan pada variabel
tentang dasar-dasar pemrograman Python. Program ini akan menampilkan
"""

#Ini merupakan membuat variabel bernama spiderman yang bersisi "peter parker"
spiderman = "peter parket"

#Ini mencetak nilai pada variabel spiderman
print(spiderman)

#mengecek type nilai pada variabel spiderman
print(type(spiderman))

#Membuat variabel secara berturut-turut
nama, alamat, jeniskelamin = "peter parker", "indonesia", "Laki-laki"

#mencetak setiap nilai pada variabel nama, alamat, dan jenis kelamin
print(nama)
print(alamat)
print(jeniskelamin)
```
