### **CASTING TYPE DATA**

Casting type data adalah proses mengubah suatu tipe data ke tipe data yang lain.

### **TO INTEGER**

```python
# float ---> integer

data = 19.9
data_int = int(data)

print(data, type(data))
print(data_int, type(data_int))
```

```python
# boolean ---> interger 

data = True
data_int = int(data)

print(data, type(data))
print(data_int, type(data_int))
```

### **TO FLOAT**

```python
# integer ---> float

data = 10
data_float = float(data)

print(data, type(data))
print(data_float, type(data_float))
```

```python
# boolean ---> float

data = True
data_float = float(data)

print(data, type(data))
print(data_float, type(data_float))

data = False
data_float = float(data)

print(data, type(data))
print(data_float, type(data_float))
```

### **TO BOOLEAN**

```python
# float/int --> boolean

data = -10
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))

#float --> boolean

data = -10.9999
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))
```

```python
# string --> boolean

data = ""
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))

data = "spiderman"
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))
```

```python
# list --> boolean

data = []
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))

data = ["spiderman"]
data_convert = bool(data)

print(data, type(data))
print(data_convert, type(data_convert))
```

### **TO BOOLEAN**

```python
# all type --> string

# integer --> string

data = 100
data_convert = str(data)

print(data, type(data))
print(data_convert, type(data_convert))
print(data, data_convert)
```
