# Business Understanding

Judul: Analisis Dataset Gaji Karyawan Berdasarkan Masa Kerja.

Latar Belakang:

Sebuah Perusahaan atau organisasi ingin mengoptimalkan pengelolaan anggaran untuk SDM dan standar gaji karyawan yang adil.dengan data historis yang mencatat hubungan antara masa kerja karyawan dan gaji yang di terimanya.

Tujuan:

* Menganalisis karakteristik data kompensasi karyawan untuk memahami pola hubungan antara masa kerja dan besaran gaji.  
* Memperoleh gambaran yang jelas mengenai sebaran kompensasi
* Mendapatkan acuan awal dalam menyusun standar penggajian yang adil dan transparan.




Masalah:

Terdapat beberapa data yang kosong. duplikat dan juga outlier

Solusi:

* Menghapus data duplikat
* Mengisi data yang kosong dengan mean,median,modus atau menghapus data
* Mengisi data outlier dengan rata - rata atau menghapusnya

Deskripsi Kolom:

* YearsExperience: Jumlah lama waktu karyawan bekerja dalam hitungan tahun.

* Salary: Jumlah gaji tahunan yang di terima oleh karyawan.
