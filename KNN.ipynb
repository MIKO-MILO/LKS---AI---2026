```python
df = df.drop(columns=['Id'])
```

```python
kolom_numerik = df.select_dtypes(include='number').columns

jumlah_kolom = 3
jumlah_baris = math.ceil(len(kolom_numerik) / jumlah_kolom)

fig, axes = plt.subplots(
    jumlah_baris,
    jumlah_kolom,
    figsize=(15, jumlah_baris * 4)
)

axes = axes.flatten()

for i, kolom in enumerate(kolom_numerik):
    sns.histplot(data=df, x=kolom, kde=True, ax=axes[i])
    axes[i].set_title(f'Histplot {kolom}')

# Menghapus subplot yang kosong
for i in range(len(kolom_numerik), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()
```
